```python
import numpy as ny
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv('insurance_pre.csv')
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
indepen=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
depen=dataset[['charges']]
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
grid_param={'n_estimators':[10,50,100],'criterion':['squared_error','friedman_mse','absolute_error','poisson'],
            'max_features':['auto','sqrt','log2']}

grid=GridSearchCV(RandomForestRegressor(),grid_param,refit=True,verbose=3,n_jobs=-1)
grid.fit(indepen,depen)
```

```python
result=grid.cv_results_
grid_prediction=grid.predict(indepen)

from sklearn.metrics import r2_score
r=r2_score(depen,grid_prediction)
print('The R_score value for best parameters :',format(grid.best_params_),r)
```

```python
table=pd.DataFrame.from_dict(result)
table
```
