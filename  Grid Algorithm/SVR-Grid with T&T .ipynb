```python
import numpy as ny
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv('insurance_pre.csv')
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
dataset.columns
```

```python
indepen=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
depen=dataset[['charges']]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(indepen,depen,test_size=1/3,random_state=0)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
grid_prameters={'kernel':['rbf','sigmoid','linear'],'C':[1,10,100,1000,2000],
                'gamma':['auto','scale']}

grid=GridSearchCV(SVR(),grid_prameters,refit=True,verbose=3,n_jobs=-1)
grid.fit(x_train,y_train)
```

```python
result=grid.cv_results_
grid_prediction=grid.predict(x_test)

from sklearn.metrics import r2_score
r=r2_score(y_test,grid_prediction)
print("The R_score value for the best parameter :",format(grid.best_params_),r)
```
